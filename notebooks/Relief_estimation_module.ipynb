```python
#Relief Estimation

import math

def estimate_relief_resources(
    exposed_population,
    relief_days=7,
    assistance_coverage=1.0,
    water_per_person=15,
    kcal_per_person=2100,
    meals_per_person=3,
    people_per_medical_kit=100
):
    if exposed_population < 0:
        raise ValueError("Population cannot be negative")

    if relief_days < 1:
        raise ValueError("Relief days must be at least 1")

    if not 0 <= assistance_coverage <= 1:
        raise ValueError("Coverage must be between 0 and 1")

    if people_per_medical_kit <= 0:
        raise ValueError("Medical kit coverage must be positive")

    people = math.ceil(
        exposed_population * assistance_coverage
    )

    return {
        "People_Planned_For": people,
        "Relief_Days": relief_days,
        "Water_Litres": people * water_per_person * relief_days,
        "Total_Food_Kcal": people * kcal_per_person * relief_days,
        "Total_Meals": people * meals_per_person * relief_days,
        "Emergency_Medical_Kits": math.ceil(
            people / people_per_medical_kit
        )
    }
```

```python
#test
result = estimate_relief_resources(
    exposed_population=10000,
    relief_days=7
)

for key, value in result.items():
    print(f"{key}: {value:,}")
```

```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd

# Load the clustered district dataset
df = pd.read_csv(
    '/content/drive/MyDrive/FloodRelief_AI/data/district_clusters.csv'
)


# Calculate relief requirements for every district
relief_results = df['Population_Exposure'].apply(
    lambda population: estimate_relief_resources(
        exposed_population=population,
        relief_days=7
    )
)

# Convert results into a DataFrame
relief_data = pd.DataFrame(relief_results.tolist())

# Combine district information with relief estimates
relief_df = pd.concat(
    [df.reset_index(drop=True), relief_data],
    axis=1
)

print("Total districts:", len(relief_df))

display(relief_df[
    [
        'State',
        'District',
        'Vulnerability_Group',
        'People_Planned_For',
        'Water_Litres',
        'Total_Food_Kcal',
        'Total_Meals',
        'Emergency_Medical_Kits'
    ]
].head(10))
```
